```python
# Load pandas for data manipulation
import pandas as pd
```

```python
# Read fund master dataset and display the first few rows
fund = pd.read_csv("../data/raw/01_fund_master.csv")
print(fund.head())
```

```python
# Show the column names of the fund master dataset
print(fund.columns)
```

```python
# Print summary statistics for the fund master dataset, including row/column counts and columns list
print("=" * 60)
print("FUND MASTER DATASET SUMMARY")
print("=" * 60)

print(f"Total Rows    : {fund.shape[0]}")
print(f"Total Columns : {fund.shape[1]}")

print("\nColumn Names:")
print(fund.columns.tolist())
```

```python
# Display unique fund houses and count how many distinct fund houses exist
print("=" * 60)
print("UNIQUE FUND HOUSES")
print("=" * 60)

print(fund["fund_house"].unique())

print("\nTotal Fund Houses:", fund["fund_house"].nunique())
```

```python
# Display unique sub-categories and count the distinct sub-category values
print("=" * 60)
print("UNIQUE SUB CATEGORIES")
print("=" * 60)

print(fund["sub_category"].unique())

print("\nTotal Sub Categories:", fund["sub_category"].nunique())
```

```python
# Display unique risk categories and count how many distinct risk categories are present
print("=" * 60)
print("UNIQUE RISK CATEGORIES")
print("=" * 60)

print(fund["risk_category"].unique())

print("\nTotal Risk Categories:", fund["risk_category"].nunique())
```

```python
# Load NAV history dataset and preview the first few records
nav = pd.read_csv("../data/raw/02_nav_history.csv")
nav.head()
```

```python
# Print the column names of the NAV history dataset
print(nav.columns)
```

```python
# Identify AMFI codes present in fund master but missing from NAV history for data consistency checks
missing_codes = fund[
    ~fund["amfi_code"].isin(nav["amfi_code"])
]
missing_codes
```

```python
# Generate an AMFI code validation report comparing fund master and NAV history datasets
print("=" * 60)
print("AMFI CODE VALIDATION REPORT")
print("=" * 60)

total_master = fund["amfi_code"].nunique()
total_nav = nav["amfi_code"].nunique()

print(f"Total Fund Master Codes : {total_master}")
print(f"Total NAV History Codes : {total_nav}")

missing_codes = fund[
    ~fund["amfi_code"].isin(nav["amfi_code"])
]

print(f"\nMissing Codes : {len(missing_codes)}")

if len(missing_codes) == 0:
    print("\n✅ SUCCESS")
    print("All AMFI codes from Fund Master exist in NAV History.")
else:
    print("\n❌ Missing Codes Found:")
    print(missing_codes[["amfi_code", "scheme_name"]])
```

```python
# Check for duplicate AMFI codes in the fund master dataset
duplicates = fund["amfi_code"].duplicated().sum()

print("=" * 60)
print("DUPLICATE CHECK")
print("=" * 60)

print(f"Duplicate AMFI Codes : {duplicates}")

if duplicates == 0:
    print("✅ No duplicate AMFI codes found.")
else:
    print("⚠ Duplicate AMFI codes exist.")
```
